# Retail Sales Analysis

```
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_excel('Retail_Sales_Dataset.xlsx')
df.head()
```

```
print(df.info())
print(df.describe())
df=df.drop_duplicates()
df['Date']=pd.to_datetime(df['Date'])
df['Month']=df['Date'].dt.strftime('%b')
```

```
df.groupby('Month')['Revenue'].sum().plot(kind='bar',title='Monthly Revenue')
plt.show()
```

```
df.groupby('Category')['Revenue'].sum().plot(kind='bar',title='Revenue by Category')
plt.show()
```

```
df.groupby('Region')['Revenue'].sum().plot(kind='bar',title='Revenue by Region')
plt.show()
```

```
df.groupby('Product')['Revenue'].sum().sort_values().plot(kind='barh',title='Top Products')
plt.show()
```

```
corr=df[['Units_Sold','Unit_Price','Revenue']].corr();plt.imshow(corr,cmap='coolwarm');plt.xticks(range(3),corr.columns,rotation=45);plt.yticks(range(3),corr.columns);plt.colorbar();plt.show()
```
